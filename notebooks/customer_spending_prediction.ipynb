## 5. Regression Model Comparison

Models: Linear Regression baseline, Multiple Linear Regression, and KNN Regressor.

Metrics: MAE, RMSE and R². Lower MAE/RMSE is better; higher R² is better.

```python
regression_metrics = pd.read_csv('../results/regression_metrics.csv')
display(regression_metrics.round(3))
print('Best regression model:', regression_metrics.sort_values('R2', ascending=False).iloc[0]['Model'])
display(Image(filename='../results/regression_comparison.png'))
display(Image(filename='../results/actual_vs_predicted.png'))
```

## 6. Classification Extension

Customers are classified as Low Spenders or High Spenders using the training-set median spending as the threshold.

Models: Logistic Regression, KNN Classifier, and Bernoulli Naive Bayes.

```python
classification_metrics = pd.read_csv('../results/classification_metrics.csv')
display(classification_metrics.round(3))
print('Best classification model:', classification_metrics.sort_values('F1', ascending=False).iloc[0]['Model'])
display(Image(filename='../results/confusion_matrix.png'))
```

## 7. Final Conclusion

Multiple Linear Regression is the strongest regression model in this experiment. Logistic Regression is the strongest classification model. The project compares multiple models and evaluates them with appropriate metrics instead of relying on a single score.
